```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, InputLayer

# Load the dataset
file_path = '/content/preprocessed data hourly.xlsx - preprocessed data hourly.xlsx.csv'
data = pd.read_csv(file_path)

# Parse the time column and set it as the index
data['time'] = pd.to_datetime(data['time'], errors='coerce')
data.set_index('time', inplace=True)

# Ensure the index is sorted
data.sort_index(inplace=True)

# Extract the first 4 months of data
data_4_months = data.loc['2011-11-11':'2012-03-10']

# Use only the temperature column for LSTM
ts = data_4_months['temperature'].values.reshape(-1, 1)

# Normalize the data
scaler = MinMaxScaler(feature_range=(0, 1))
ts_scaled = scaler.fit_transform(ts)

# Split into training and testing sets
train_size = int(len(ts_scaled) * 0.8)
train, test = ts_scaled[:train_size], ts_scaled[train_size:]

# Create sequences
def create_sequences(data, seq_length):
    xs, ys = [], []
    for i in range(len(data) - seq_length):
        x = data[i:i+seq_length]
        y = data[i + seq_length]
        xs.append(x)
        ys.append(y)
    return np.array(xs), np.array(ys)

seq_length = 24  # Example sequence length
X_train, y_train = create_sequences(train, seq_length)
X_test, y_test = create_sequences(test, seq_length)

# Build the LSTM model
model = Sequential()
model.add(InputLayer(input_shape=(seq_length, 1)))
model.add(LSTM(50, return_sequences=True))
model.add(LSTM(50, return_sequences=False))
model.add(Dense(25))
model.add(Dense(1))

model.compile(optimizer='adam', loss='mean_squared_error')

# Train the model
model.fit(X_train, y_train, batch_size=1, epochs=1)

# Make predictions
predictions = model.predict(X_test)
predictions = scaler.inverse_transform(predictions)

# Plot the results
train_plot = np.empty_like(ts_scaled)
train_plot[:, :] = np.nan
train_plot[seq_length:len(train) + seq_length, :] = train

test_plot = np.empty_like(ts_scaled)
test_plot[:, :] = np.nan
test_plot[len(train) + seq_length - 1:len(train) + seq_length + len(predictions) - 1, :] = predictions

plt.figure(figsize=(10, 6))
plt.plot(scaler.inverse_transform(ts_scaled), label='True Value')
plt.plot(train_plot, label='Train')
plt.plot(test_plot, label='Predicted')
plt.legend()
plt.title('LSTM Model Forecast')
plt.show()
```

```python
# Install required packages
!pip install dash
!pip install dash-core-components
!pip install dash-html-components
!pip install dash-bootstrap-components
!pip install plotly

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import LabelEncoder
import plotly.express as px
import dash
from dash import dcc, html
from dash.dependencies import Input, Output

# Load the dataset
file_path = '/content/preprocessed data hourly.xlsx - preprocessed data hourly.xlsx.csv'
data = pd.read_csv(file_path)

# Parse the time column and set it as the index
data['time'] = pd.to_datetime(data['time'], errors='coerce')
data.set_index('time', inplace=True)

# Ensure the index is sorted
data.sort_index(inplace=True)

# Extract the first 4 months of data
data_4_months = data.loc['2011-11-11':'2012-03-10']

# Drop rows with missing values
data_4_months = data_4_months.dropna()

# Encode categorical features
label_encoder = LabelEncoder()
data_4_months['summary'] = label_encoder.fit_transform(data_4_months['summary'])
data_4_months['icon'] = label_encoder.fit_transform(data_4_months['icon'])

# Define the features and target
features = data_4_months.drop(columns=['temperature'])
target = data_4_months['temperature']

# Ensure all features are numeric
features = pd.get_dummies(features)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=42)

# Train the Random Forest Regressor
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Get feature importances
importances = model.feature_importances_
feature_names = features.columns

# Plot feature importances
plt.figure(figsize=(10, 6))
plt.barh(feature_names, importances, color='green')
plt.xlabel('Feature Importance')
plt.title('Feature Importance Using Random Forest')
plt.show()

# Create a simple dashboard using Dash
app = dash.Dash(__name__)

app.layout = html.Div([
    html.H1('Energy Consumption Forecast Dashboard'),
    dcc.Graph(id='feature-importance'),
    dcc.Graph(id='consumption-forecast'),
])

@app.callback(
    Output('feature-importance', 'figure'),
    Input('feature-importance', 'id')
)
def update_feature_importance(_):
    fig = px.bar(
        x=importances,
        y=feature_names,
        orientation='h',
        labels={'x': 'Feature Importance', 'y': 'Features'},
        title='Feature Importance Using Random Forest'
    )
    return fig

# Assuming you have predictions from ARIMA and LSTM
# Here, I'm using dummy data for the forecast graph
@app.callback(
    Output('consumption-forecast', 'figure'),
    Input('consumption-forecast', 'id')
)
def update_forecast_graph(_):
    dates = pd.date_range(start='2012-03-11', periods=30, freq='D')
    forecast_data = np.random.rand(30)  # Dummy forecast data
    fig = px.line(
        x=dates, y=forecast_data,
        labels={'x': 'Date', 'y': 'Predicted Energy Consumption'},
        title='Energy Consumption Forecast'
    )
    return fig

if __name__ == '__main__':
    app.run_server(debug=True)

```
